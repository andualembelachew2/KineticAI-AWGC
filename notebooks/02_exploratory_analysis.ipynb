# 02 Exploratory Analysis

This notebook explores the AWGC phase-evolution datasets using basic visualization and statistical regression.

The goal is to understand how sintering temperature and SBF soaking time influence phase composition, network connectivity, densification, and degradation trends.

## Kinetic Blueprint Dataset Exploration

This section explores relationships among phase architecture, network connectivity, transport behavior, and bioactivity using the Kinetic Blueprint Master Dataset.

Research Question:

How do phase composition, density, and NBO/BO ratio govern transport behavior and degradation response in glass-ceramic systems?

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress
from pathlib import Path

ROOT_DIR = Path("..")
DATA_DIR = ROOT_DIR / "data"

kb_df = pd.read_csv(DATA_DIR / "kinetic_blueprint_master.csv")

def plot_with_regression(x, y, xlabel, ylabel, title, temperatures=None):
    slope, intercept, r_value, p_value, std_err = linregress(x, y)

    plt.figure(figsize=(7,5))
    plt.scatter(x, y, s=120)

    if temperatures is not None:
        for x_val, y_val, temp in zip(x, y, temperatures):
            plt.annotate(f"{int(temp)}°C", (x_val, y_val))

    x_fit = np.linspace(min(x), max(x), 100)
    y_fit = intercept + slope * x_fit

    plt.plot(x_fit, y_fit, linestyle='--')

    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.title(title)

    plt.text(
        0.05,
        0.95,
        f"R² = {r_value**2:.3f}\nr = {r_value:.3f}\np = {p_value:.3e}",
        transform=plt.gca().transAxes,
        verticalalignment='top'
    )

    plt.tight_layout()
    plt.show()

    print(f"Slope = {slope:.4f}")
    print(f"R² = {r_value**2:.4f}")
    print(f"Pearson r = {r_value:.4f}")
    print(f"p-value = {p_value:.4e}")

kb_df.head()
```

## Relationship Between Network Connectivity and Degradation

### Scientific Motivation

The Kinetic Blueprint framework proposes that network connectivity governs transport behavior in glass-ceramic systems.

The NBO/BO ratio serves as a structural descriptor of network depolymerization.

Hypothesis:

- Higher NBO/BO ratio → less connected glass network
- Less connected network → higher transport activity
- Higher transport activity → greater mass loss and ion release

The following analysis evaluates the relationship between NBO/BO ratio and degradation behavior after 21 days of SBF immersion with statistical linear regression.

```python
plot_with_regression(
    kb_df["NBO_BO_Ratio"],
    kb_df["MassLoss_21d_Percent"],
    "NBO/BO Ratio",
    "Mass Loss after 21 Days (%)",
    "Network Connectivity vs Degradation Behavior",
    temperatures=kb_df["Temperature_C"]
)
```

## Initial Interpretation

A clear positive relationship is expected between NBO/BO ratio and mass loss.

Samples with higher NBO/BO ratios possess more depolymerized glass networks and a greater concentration of non-bridging oxygens, which facilitate water penetration and network dissolution.

Consequently:

- Higher NBO/BO ratio → higher degradation
- Higher NBO/BO ratio → greater ion release
- Lower NBO/BO ratio → more condensed networks
- More condensed networks → diffusion-governed transport

This trend represents a central hypothesis of the Kinetic Blueprint framework, linking network connectivity to transport behavior and bio-interface performance.

## Relationship Between Network Connectivity and Calcium Release

### Scientific Motivation

Calcium release is a key indicator of glass-network dissolution and microenvironmental engineering.

If the Kinetic Blueprint hypothesis is correct, NBO/BO ratio should influence not only degradation behavior but also ionic transport.

The following analysis evaluates the relationship between network connectivity and calcium release after 21 days of SBF immersion.

```python
plot_with_regression(
    kb_df["NBO_BO_Ratio"],
    kb_df["Ca_Release_21d_mM"],
    "NBO/BO Ratio",
    "Calcium Release (mM)",
    "Network Connectivity vs Calcium Release",
    temperatures=kb_df["Temperature_C"]
)
```

## Interpretation of Calcium Release Behavior

A positive relationship is expected between NBO/BO ratio and calcium release.

Samples with higher NBO/BO ratios contain more depolymerized glass networks, increasing susceptibility to dissolution and facilitating ionic transport into the surrounding medium.

The observed trend suggests that network connectivity influences not only bulk degradation but also calcium-mediated microenvironmental modification.

Together with the mass-loss analysis, these results support the central Kinetic Blueprint hypothesis that transport behavior can be programmed through control of glass-network structure.

## Relationship Between Network Connectivity and Silicon Release

### Scientific Motivation

Silicon release reflects dissolution of the silicate network itself and therefore provides a direct measure of network degradation.

Unlike calcium, which acts primarily as a network modifier, silicon originates from the glass framework.

The following analysis evaluates whether NBO/BO ratio governs silicon transport behavior after 21 days of SBF immersion.

```python
plot_with_regression(
    kb_df["NBO_BO_Ratio"],
    kb_df["Si_Release_21d_mM"],
    "NBO/BO Ratio",
    "Silicon Release (mM)",
    "Network Connectivity vs Silicon Release",
    temperatures=kb_df["Temperature_C"]
)
```

## Relationship Between Densification and Degradation

### Scientific Motivation

Thermal processing influences not only network connectivity but also densification.

Increasing density may reduce open transport pathways and limit fluid penetration.

This analysis evaluates whether bulk density acts as a physical barrier to degradation and transport in glass-ceramic systems.

```python
plot_with_regression(
    kb_df["Bulk_Density_g_cm3"],
    kb_df["MassLoss_21d_Percent"],
    "Bulk Density (g/cm³)",
    "Mass Loss after 21 Days (%)",
    "Densification vs Degradation Behavior",
    temperatures=kb_df["Temperature_C"]
)
```

### Scientific Motivation

Phase architecture represents the structural consequence of thermal processing.
