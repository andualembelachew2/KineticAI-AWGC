## Kinetic Blueprint Dataset Exploration

This section explores relationships among phase architecture, network connectivity, transport behavior, and bioactivity using the Kinetic Blueprint Master Dataset.

Research Question:

How do phase composition, density, and NBO/BO ratio govern transport behavior and degradation response in glass-ceramic systems?

```python
import pandas as pd
from pathlib import Path

ROOT_DIR = Path("..")
DATA_DIR = ROOT_DIR / "data"

kb_df = pd.read_csv(DATA_DIR / "kinetic_blueprint_master.csv")

kb_df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))

plt.scatter(
    kb_df["NBO_BO_Ratio"],
    kb_df["MassLoss_21d_Percent"],
    s=120
)

for _, row in kb_df.iterrows():
    plt.annotate(
        f"{int(row['Temperature_C'])}°C",
        (row["NBO_BO_Ratio"], row["MassLoss_21d_Percent"])
    )

plt.xlabel("NBO/BO Ratio")
plt.ylabel("Mass Loss after 21 Days (%)")
plt.title("Network Connectivity vs Degradation Behavior")

plt.tight_layout()
plt.show()
```
