## 07 — XRD Pattern Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Paths
ROOT_DIR = Path("..")
DATA_DIR = ROOT_DIR / "data" / "raw"
FIGURE_DIR = ROOT_DIR / "figures" / "characterization"

# Ensure figure directory exists
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

# Load dataset
xrd_df = pd.read_csv(DATA_DIR / "xrd_patterns.csv")
display(xrd_df.head())

# Plot representative patterns
plt.figure(figsize=(10, 6))

for temp in [700, 900, 1100]:
    plt.plot(
        xrd_df["2Theta_deg"],
        xrd_df[f"Intensity_{temp}C"],
        label=f"{temp}°C"
    )

plt.xlabel("2θ (degrees)")
plt.ylabel("Intensity (a.u.)")
plt.title("XRD Phase Evolution")
plt.legend()
plt.tight_layout()

# Save figure
output_path = FIGURE_DIR / "xrd_phase_evolution.png"
plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# Check if file exists
print(f"File generated: {output_path.exists()}")
```
