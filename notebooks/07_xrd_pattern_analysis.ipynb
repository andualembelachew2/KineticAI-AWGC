## 07 — XRD Pattern Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Load dataset
xrd_df = pd.read_csv("../data/raw/xrd_patterns.csv")

display(xrd_df.head())

# Create output folder if needed
Path("../figures/characterization").mkdir(
    parents=True,
    exist_ok=True
)

# Plot representative patterns
plt.figure(figsize=(10,6))

for temp in [700, 900, 1100]:
    plt.plot(
        xrd_df["2Theta_deg"],
        xrd_df[f"Intensity_{temp}C"],
        label=f"{temp}°C"
    )

plt.xlabel("2θ (degrees)")
plt.ylabel("Intensity (a.u.)")
plt.title("XRD Phase Evolution")
plt.legend()
plt.tight_layout()

# Save figure
plt.savefig(
    "../figures/characterization/xrd_phase_evolution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Saved: ../figures/characterization/xrd_phase_evolution.png")
```
