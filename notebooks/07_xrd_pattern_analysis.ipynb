## Crystallization Pathways and the Kinetic Blueprint Framework

The raw XRD dataset represents the earliest experimental evidence within the Kinetic Blueprint framework.

Thermal processing governs crystallization behavior, which determines phase architecture and subsequently influences transport behavior, microenvironment evolution, and biological response.

Research Question:

How does thermal processing regulate crystallization pathways in spray-pyrolyzed apatite–wollastonite glass-ceramic systems?

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

ROOT_DIR = Path("..")
DATA_DIR = ROOT_DIR / "data"

xrd_df = pd.read_csv(DATA_DIR / "xrd_patterns.csv")

display(xrd_df.head())

plt.figure(figsize=(10,6))

temperatures = [700, 800, 900, 1000, 1100]

for temp in temperatures:
    plt.plot(
        xrd_df["2Theta_deg"],
        xrd_df[f"Intensity_{temp}C"],
        label=f"{temp}°C"
    )

plt.xlabel("2θ (degrees)")
plt.ylabel("Intensity (a.u.)")
plt.title("XRD Patterns as a Function of Sintering Temperature")

plt.legend()
plt.tight_layout()

plt.savefig(
    "../figures/characterization/xrd_phase_evolution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

## Initial Interpretation

The XRD patterns reveal the evolution of crystallization behavior across the processing window.

### 700 °C
- Broad amorphous contribution
- Limited crystallization

### 800–1000 °C
- Progressive peak development
- Increasing crystallinity

### 1100 °C
- Strong crystalline reflections
- Maximum phase development

These observations support the hypothesis that thermal processing programs phase architecture through crystallization pathways.

## Crystallization Evolution

### Scientific Motivation

The XRD patterns indicate progressive crystallization with increasing sintering temperature.

Quantifying peak development provides a simple proxy for crystallization evolution and links the raw diffraction data to the phase-fraction datasets used throughout the Kinetic Blueprint framework.

The following analysis tracks the intensity of the dominant crystalline reflection as a function of processing temperature.

```python
peak_region = xrd_df[
    (xrd_df["2Theta_deg"] >= 29.5) &
    (xrd_df["2Theta_deg"] <= 30.5)
]

peak_summary = []

for temp in [700, 800, 900, 1000, 1100]:
    peak_intensity = peak_region[f"Intensity_{temp}C"].max()
    peak_summary.append({
        "Temperature_C": temp,
        "Peak_Intensity": peak_intensity
    })

peak_summary_df = pd.DataFrame(peak_summary)

plt.figure(figsize=(7,5))

plt.plot(
    peak_summary_df["Temperature_C"],
    peak_summary_df["Peak_Intensity"],
    marker="o",
    linewidth=2
)

plt.xlabel("Sintering Temperature (°C)")
plt.ylabel("Peak Intensity (a.u.)")
plt.title("Crystallization Evolution with Temperature")

plt.tight_layout()
plt.show()

peak_summary_df
```

## Interpretation

Peak intensity increases systematically with sintering temperature, indicating progressive crystallization of the glass-ceramic system.

These observations indicate that thermal processing systematically alters crystallization behavior, leading to measurable changes in phase architecture. This provides the experimental foundation linking processing conditions to subsequent transport behavior and microenvironment evolution.
