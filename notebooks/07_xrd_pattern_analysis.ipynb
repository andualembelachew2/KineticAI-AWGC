## XRD Pattern & Phase Evolution Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Paths
ROOT_DIR = Path("..")
DATA_DIR = ROOT_DIR / "data" / "raw"
FIGURE_DIR = ROOT_DIR / "figures" / "characterization"

# Create figure directory if it doesn't exist
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

# Load XRD dataset
xrd_df = pd.read_csv(DATA_DIR / "xrd_patterns.csv")

display(xrd_df.head())

# Plot representative patterns
plt.figure(figsize=(10,6))

for temp in [700, 900, 1100]:
    plt.plot(
        xrd_df["2Theta_deg"],
        xrd_df[f"Intensity_{temp}C"],
        label=f"{temp}°C",
        linewidth=1.5
    )

plt.xlabel("2θ (degrees)")
plt.ylabel("Intensity (a.u.)")
plt.title("XRD Phase Evolution")
plt.legend()
plt.tight_layout()

# Save figure
output_file = FIGURE_DIR / "xrd_phase_evolution.png"

plt.savefig(
    output_file,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print(f"Figure saved to: {output_file}")
```
