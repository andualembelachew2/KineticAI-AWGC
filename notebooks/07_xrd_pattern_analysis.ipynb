## 07 — XRD-Derived Phase Composition Analysis

This notebook analyzes the crystallization and phase-architecture evolution of spray-pyrolyzed apatite–wollastonite glass-ceramics using the XRD-quantified phase fractions reported in Workie et al., *Crystals* 13(7):1049 (2023) and the Kinetic Blueprint studies.

The raw diffraction spectra are published with the source articles; this repository ships the quantified phase fractions extracted from them (see `docs/publications.md`).

## Research Question

How does thermal processing (700–1100 °C) reshape the phase architecture of the material?

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

ROOT_DIR = Path("..")
DATA_DIR = ROOT_DIR / "data"
FIGURE_DIR = ROOT_DIR / "figures" / "characterization"
FIGURE_DIR.mkdir(parents=True, exist_ok=True)

kb_df = pd.read_csv(DATA_DIR / "kinetic_blueprint_master.csv")
phase_cols = ["Amorphous_Percent", "Wollastonite_Percent", "Hydroxyapatite_Percent", "Whitlockite_Percent"]
kb_df[["Temperature_C"] + phase_cols]
```

## Phase Architecture vs Sintering Temperature

Each row is an XRD-quantified phase composition; the four phase fractions sum to 100 %.

```python
plt.figure(figsize=(9, 6))

for col in phase_cols:
    plt.plot(
        kb_df["Temperature_C"],
        kb_df[col],
        marker="o",
        label=col.replace("_Percent", "")
    )

plt.xlabel("Sintering temperature (°C)")
plt.ylabel("Phase fraction (%)")
plt.title("XRD-Derived Phase Composition vs Sintering Temperature")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()

output_path = FIGURE_DIR / "xrd_phase_evolution.png"
plt.savefig(output_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Figure saved: {output_path.resolve()}")
```

## Crystallization Trends

Crystallinity is the complement of the residual amorphous fraction:

```python
kb_df["Crystallinity_Percent"] = 100 - kb_df["Amorphous_Percent"]

summary = kb_df[["Temperature_C", "Amorphous_Percent", "Crystallinity_Percent"]].copy()
summary.columns = ["Temperature (°C)", "Amorphous (%)", "Crystallinity (%)"]
summary
```

## Interpretation

- Amorphous content falls from **52.2 % (700 °C)** to **22.6 % (1100 °C)**, while crystallinity rises correspondingly.
- The crystalline architecture shifts from hydroxyapatite-dominated (700 °C) toward wollastonite + whitlockite-dominated (1100 °C).
- This phase partitioning is the structural foundation of the transport and bioresponse behavior analyzed in notebooks 02, 05, 06, and 09.

```python
print("Phase fractions sum to 100 %:", all(abs(kb_df[phase_cols].sum(axis=1) - 100) < 0.05))
print(f"Amorphous content: {kb_df['Amorphous_Percent'].max():.1f}% (700°C) -> {kb_df['Amorphous_Percent'].min():.1f}% (1100°C)")
```
