## Crystallization Pathways and the Kinetic Blueprint Framework

The raw XRD dataset represents the earliest experimental evidence within the Kinetic Blueprint framework.

Thermal processing governs crystallization behavior, which determines phase architecture and subsequently influences transport behavior, microenvironment evolution, and biological response.

Research Question:

How does thermal processing regulate crystallization pathways in spray-pyrolyzed apatite–wollastonite glass-ceramic systems?

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

ROOT_DIR = Path("..")
DATA_DIR = ROOT_DIR / "data"

xrd_df = pd.read_csv(DATA_DIR / "xrd_patterns.csv")

display(xrd_df.head())

plt.figure(figsize=(10,6))

temperatures = [700, 800, 900, 1000, 1100]

for temp in temperatures:
    
    plt.plot(
        xrd_df["2Theta_deg"],
        xrd_df[f"Intensity_{temp}C"],
        label=f"{temp}°C"
    )

plt.xlabel("2θ (degrees)")
plt.ylabel("Intensity (a.u.)")
plt.title("XRD Patterns as a Function of Sintering Temperature")

plt.legend()
plt.tight_layout()
plt.show()
```

## Initial Interpretation

The XRD patterns reveal the evolution of crystallization behavior across the processing window.

### 700 °C
- Broad amorphous contribution
- Limited crystallization

### 800–1000 °C
- Progressive peak development
- Increasing crystallinity

### 1100 °C
- Strong crystalline reflections
- Maximum phase development

These observations support the hypothesis that thermal processing programs phase architecture through crystallization pathways.
