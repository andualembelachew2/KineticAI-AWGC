# Exploratory Phase Evolution Analysis

This notebook inspects the phase composition dataset across thermal treatment conditions, summarizing temperature boundaries and tracking amorphous fraction limits.

```python
import pandas as pd

# Load phase composition dataset
df = pd.read_csv("../data/processed/phase_composition.csv")
df.head()
```

```python
# Summary statistics for all numeric columns
df.describe()
```

```python
# Extract temperature range and amorphous content extremes
print("Temperature Range:")
print(df["Temperature"].min(), "to", df["Temperature"].max(), "°C")

print("\nMaximum Amorphous Content:")
print(df["Amorphous"].max())

print("\nMinimum Amorphous Content:")
print(df["Amorphous"].min())
```
