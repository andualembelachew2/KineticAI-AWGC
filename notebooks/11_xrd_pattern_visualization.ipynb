### Module 07 — XRD Pattern & Phase Evolution Analysis

Loads raw diffraction data across thermal processing regimes ($700^\circ\text{C}$ vs $1100^\circ\text{C}$) to track crystallization and phase architecture transformations.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load raw XRD dataset
df = pd.read_csv("../data/raw/xrd_patterns.csv")
display(df.head())

# Plotting Phase Evolution
plt.figure(figsize=(10, 6))

plt.plot(df["2Theta_deg"], df["Intensity_700C"], label="700°C")
plt.plot(df["2Theta_deg"], df["Intensity_900C"], label="900°C")
plt.plot(df["2Theta_deg"], df["Intensity_1100C"], label="1100°C")

plt.xlabel("2θ (degrees)")
plt.ylabel("Intensity (a.u.)")
plt.title("XRD Phase Evolution")
plt.legend()

# Save figure before show
plt.savefig(
    "../figures/characterization/xrd_phase_evolution.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
